## 2. Laboratorio Matemático: Azar vs. Inversión con Interés Compuesto

### Fundamento de Probabilidad Colombiana:
* **Baloto:** Para acertar los 5 números (del 1 al 43) y la súper balota (del 1 al 16):
  $$\Omega = \binom{43}{5} \times 16 = 962.598 \times 16 = 15.401.568 \implies P(\text{Ganar}) = \frac{1}{15.401.568} \approx 0.0000065\%$$
* **Chance de 4 cifras:** Cuatro dígitos independientes del 0 al 9:
  $$\Omega = 10^4 = 10.000 \implies P(\text{Ganar}) = \frac{1}{10.000} = 0.01\%$$
* **Esperanza Matemática:** En los juegos de suerte y azar en Colombia, el retorno teórico al público (RTP) oscila entre el 40% y el 50%. En promedio, por cada $1.000 COP apostados, el jugador pierde entre $500 y $600 COP ($E[X] - \text{Costo} < 0$).

```python
# Comparativa: $200.000 COP mensuales a 10 años en azar vs. Inversión al 11% EA
monto_mensual = 200_000.0
anios = 10
tasa_ea = 0.11

resultado_comparativa = comparar_azar_vs_inversion_compuesta(
    gasto_mensual_juegos_azar_cop=monto_mensual,
    horizonte_anios=anios,
    tasa_ea_inversion_regulada=tasa_ea,
)

print('=== COMPARATIVA A 10 AÑOS ===')
print(f'Gasto mensual evaluado: ${monto_mensual:,.0f} COP')
print(f'Total desembolsado en 10 años: ${resultado_comparativa["total_desembolsado_cop"]:,.0f} COP')
print(f'Probabilidad Baloto: {resultado_comparativa["probabilidad_combinatoria_baloto"]}')
print(f'Probabilidad Chance 4 cifras: {resultado_comparativa["probabilidad_chance_4_cifras"]}')
print(f'Valor Esperado Teórico en Azar: ${resultado_comparativa["valor_esperado_azar_cop"]:,.0f} COP')
print(f'Pérdida Matemática en Azar: ${resultado_comparativa["perdida_matematica_esperada_azar_cop"]:,.0f} COP')
print(f'Patrimonio en Inversión Compuesta (11% EA): ${resultado_comparativa["patrimonio_inversion_compuesta_cop"]:,.0f} COP')
print(f'Intereses Compuestos Ganados: ${resultado_comparativa["ganancia_intereses_compuestos_cop"]:,.0f} COP')
print(f'BRECHA PATRIMONIAL A FAVOR DEL AHORRO: ${resultado_comparativa["brecha_riqueza_inversion_vs_azar_cop"]:,.0f} COP')
```

```python
# Generación y visualización de la trayectoria cronológica mes a mes
trayectoria = simular_trayectoria_mensual_azar_vs_inversion(monto_mensual, anios=anios, tasa_ea_inversion=tasa_ea)

print(f'{"Mes":>4} | {"Año":>4} | {"Aportado (COP)":>15} | {"Esperado Azar":>15} | {"Inversión (COP)":>16} | {"Brecha (COP)":>15}')
print('-' * 78)
for punto in trayectoria:
    print(f'{punto["mes"]:4d} | {punto["anio"]:4.1f} | ${punto["total_aportado_cop"]:14,.0f} | ${punto["valor_esperado_azar_cop"]:14,.0f} | ${punto["patrimonio_inversion_cop"]:15,.0f} | ${punto["brecha_patrimonial_cop"]:14,.0f}')
```

## 3. Detector Matemático de Pirámides y Captación Ilegal (Art. 316 Código Penal)

Bajo el **Artículo 316 del Código Penal Colombiano**, la captación masiva y habitual de dineros sin autorización de la SFC o Supersociedades acarrea penas de prisión.

Matemáticamente, todo esquema piramidal (Ponzi) que promete rentabilidades exorbitantes 'fijas y seguras' (ej. 10% mensual) requiere una progresión geométrica de reclutamiento:
$$N_k = r^k$$
donde $r$ es el número de personas que cada participante debe vincular. En muy pocos escalones, el número de aportantes necesarios supera la **población adulta total de Colombia (39.5 millones)** y la **población mundial (8.050 millones)**, provocando el colapso matemático ineludible.

```python
# Simulación: Promesa del 10% mensual fijo con reclutamiento de 3 personas por nivel
promesa_mensual = 10.0
personas_reclutadas = 3

alerta_ponzi = detectar_esquema_piramidal_ponzi(
    rentabilidad_mensual_prometida_pct=promesa_mensual,
    personas_reclutadas_por_ciclo=personas_reclutadas,
)

print('=== ANÁLISIS DE CAPTACIÓN ILEGAL ===')
print(f'Rentabilidad prometida: {alerta_ponzi["tasa_mensual_prometida_pct"]}% EM')
print(f'Tasa Efectiva Anual equivalente: {alerta_ponzi["tasa_ea_equivalente_pct"]:,.1f}% EA')
print(f'¿Alerta de Pirámide?: {alerta_ponzi["alerta_piramide"]}')
print(f'Niveles hasta agotar población adulta de Colombia: {alerta_ponzi["ciclos_para_agotar_poblacion_colombia"]} niveles\n')

print(f'{"Nivel":>6} | {"Personas Requeridas en este Nivel":>35} | {"Comparativa con Colombia":>30}')
print('-' * 76)
for nivel in alerta_ponzi['progresion_geometrica']:
    k = nivel['nivel_piramide']
    n = nivel['personas_requeridas']
    pct_colombia = (n / POBLACION_ADULTA_COLOMBIA) * 100
    comp = f'{pct_colombia:.1f}% de adultos Col.' if n < POBLACION_ADULTA_COLOMBIA else '¡SUPERA A COLOMBIA!'
    print(f'{k:6d} | {n:35,d} | {comp:>30}')
```
